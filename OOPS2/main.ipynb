```python
class Bag:
    def __init__(self,material,zip):
        self.material=material
        self.zip=zip
    def show(self):
        print(f"things inside is {self.material},{self.zip}")

rebok = Bag("Leather",3)
campus = Bag("Leather",2)
rebok.show()
campus.show()
```

# **Method**

```python
class Animal:
    def __init__(self,age):
        self.age=age
    def show(self): # instance method 
        print(f"The age is {self.age}")
    @classmethod
    def hello(cls):
        print(f"Hello Brother how are you{cls}")
    @staticmethod
    def bro():
        print("Hello Bro how are you")

dog = Animal(12)
dog.show()
dog.hello()
dog.bro()
```

# **Inheritance**

```python
class Animal:
    def __init__(self,name):
        self.name=name
    def show(self):
        print(f"Name is {self.name}")

class Human(Animal):
    def __init__(self, name,age):
        super().__init__(name)
        self.age=age
    def show(self):
            print(f"Name is {self.name} and age is {self.age}")

h1 = Human("Subhajit",21)
a1 =Animal("Lion")
h1.show()
a1.show()

```

```python
## Multiple Inheritance 

class Grandpa:
    def __init__(self,name):
        self.name=name
    
class Papa(Grandpa):
    def __init__(self, name,age):
        super().__init__(name)
        self.age = age
class Me(Papa):
    def __init__(self, name, age,blood):
        super().__init__(name, age)
        self.blood=blood
    def show(self):
        print(f"{self.name},{self.age},{self.blood}")

Subhajit=Me('Subhajit',21,'B+')
Subhajit.show()
```
