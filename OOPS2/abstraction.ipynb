```python
from abc import ABC,abstractmethod

class Abstrction:

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def peremiter(self):
        pass

class square(Abstrction):
    def __init__(self,side):
        self.side=side
        
    def area(self):
        print("This is suare area ")

    def peremiter(self):
        print("This is square paremiter ")

class Circle(Abstrction):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        print("This is Circle area ")

    def peremiter(self):
        print("This is Circle paremiter ")

obj1=square(15)
obj2=Circle(12)

obj1.area()
```
